# QML Intrusion Detection — training & zero-day evaluation
Pre-trains every model for each held-out attack family and caches them in `models/cache/` for the dashboard.

```
%pip install -q -r ../requirements.txt
```

```
import sys, os; sys.path.insert(0, '..')
import pandas as pd
from qml.data import synthetic, load_csv
from qml.experiment import run
from qml.models import REGISTRY
path = '../data/sample/flows.csv'
df, key = (load_csv(path), 'real') if os.path.exists(path) else (synthetic(), 'demo')
print(key, df.shape); df.family.value_counts()
```

```
rows = []
for h in ['dos', 'probe', 'r2l', 'u2r']:
    res = run(df, holdout=h, cache_key=key)
    for n, r in res.items(): rows.append({'holdout': h, 'model': n, **r['metrics']})
results = pd.DataFrame(rows); results.round(3)
```

```
results.pivot(index='model', columns='holdout', values='Zero-day detection').round(3)
```

```
results.to_csv('../results_zero_day.csv', index=False)
```
